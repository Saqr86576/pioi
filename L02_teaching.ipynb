## 1) القوائم (Lists) — الإنشاء

<div dir="rtl">

<b>قائمة (list)</b>: حاوية ليها ترتيب، بين قوسين مربعين، وبتحمّل أي نوع من الأنواع.
</div>

```python
# Single values from Lecture 1 — fine for one item each
drink = "Latte"
price = 95.5

# A list holds MANY values in one variable
drinks = ["Latte", "Tea", "Mocha"]
prices = [95.5, 40.0, 55.0]
print(drinks)
print(type(drinks))     # <class 'list'>
print(len(drinks))      # 3
```

## 2) العدّ والتقطيع — نفس قواعد النصوص بالظبط

<div dir="rtl">
النص اللي اتعلمناه في L1 بيشتغل هنا بنفس الشكل: <b>العدّ من صفر</b>، والسالب من
الآخر، و <b>slicing آخره مش متاخد</b>. الفرق الوحيد إن العناصر بقى ليها أي نوع.
</div>

```python
drinks = ["Latte", "Tea", "Mocha"]

print(drinks[0])        # Latte
print(drinks[-1])       # Mocha
print(drinks[0:2])      # ['Latte', 'Tea']
print(drinks[:1])       # ['Latte']
print(drinks[1:])       # ['Tea', 'Mocha']
```

## 3) أوامر جاهزة على القائمة (methods)

<div dir="rtl">
الصيغة زي النص بالظبط: <code>القائمة.method()</code> — بس دي <b>بتعدّل</b> القائمة
نفسها مكانها. تعالى نبني سلة مشتريات ونشتغل على أوامرها:
</div>

```python
cart = []
cart.append("Latte")        # add to the end
cart.append("Croissant")
cart.append("Tea")
print(cart)                 # ['Latte', 'Croissant', 'Tea']

cart.remove("Croissant")    # remove the FIRST item with that value
cart.insert(0, "Water")     # add at a specific position
print(cart)                 # ['Water', 'Latte', 'Tea']

last = cart.pop()           # remove AND return the last item
print(last)                 # Tea
print(cart)                 # ['Water', 'Latte']

cart.sort()                 # sort in place
print(cart)                 # ['Latte', 'Water']
print(len(cart))            # 2
```

## 4) mutable + فخّ التشابه (aliasing)

<div dir="rtl">
نقطة جوهرية: الـ list <b>قابلة للتغيير (mutable)</b> — عكس النص. وعشان كده فيه فخّ
اسمها <b>aliasing</b>: لما تكتب <code>b = a</code> على قائمة، <b>b مش نسخة</b> — ده
اسم تاني لنفس القائمة، وأي تعديل فيهم يظهر في الاتنين. النسخة الحقيقية:
<code>a.copy()</code>.
</div>

```python
a = [1, 2, 3]
b = a                 # NOT a copy — two names for the SAME list
b.append(4)
print("a =", a)       # a = [1, 2, 3, 4]  ← اتغيّرت معاها! ده الفخ
print("b =", b)       # b = [1, 2, 3, 4]

c = a.copy()          # a REAL copy — independent
c.append(5)
print("a =", a)       # a = [1, 2, 3, 4]  ← زي ما هي
print("c =", c)       # c = [1, 2, 3, 4, 5]
```

```python
# The aliasing trap in one line — then the fix
prices_a = [10, 20, 30]
prices_b = prices_a       # BUG on purpose: same list
prices_b.append(40)
print("prices_a:", prices_a)   # [10, 20, 30, 40] — مش المفروض!

prices_b = prices_a.copy()     # THE FIX: one word — copy()
print("prices_a:", prices_a)   # هتفضل زي ما هي لو عدّلت قبل الـ append
```

## 5) القواميس (Dictionaries) — الإنشاء والوصول

<div dir="rtl">
الـ dict كل عنصر فيه <b>مفتاح (key)</b> و <b>قيمة (value)</b> — بين قوسين معروفين
<code>{ }</code>. ده بالظبط شكل سجل واحد من كارت مشروعك في L1: الحقل = المفتاح،
والقيمة = القيمة.
</div>

```python
record = {
    "date": "2026-09-25",
    "category": "food",
    "amount_egp": 85.5,
    "payment_method": "cash",
    "quantity": 1,
}

print(record)
print(type(record))          # <class 'dict'>
print(record["category"])    # food   — الوصول بالمفتاح بين قوسين مربعين
print(record["amount_egp"])  # 85.5
```

## 6) dict: إضافة وتعديل وحذف + [] مقابل .get()

<div dir="rtl">
نفس صيغة الإضافة والتعديل: <code>d[key] = value</code> — لو المفتاح موجود بيتعدّل،
ولو مش موجود بيتعمل. والحذف بـ <code>del d[key]</code>.<br><br>
<b>وحاجة مهمة:</b> لو دوّرت على مفتاح مش موجود بـ <code>[]</code> هيرمي
<code>KeyError</code> — عشان كده فيه <code>.get(key, default)</code> اللي بيرجّع
الـ default وبيكمّل عادي. نشوف الاتنين:
</div>

```python
record["quantity"] = 2       # update (or add if missing)
record["note"] = "paid"      # brand-new key
print(record["note"])

del record["note"]           # delete by key
print("note" in record)      # False
```

```python
# [] on a missing key raises KeyError — caught live so we can READ the message
try:
    print(record["tips"])
except KeyError as e:
    print("KeyError:", e)

# .get() answers quietly with the default
print(record.get("tips", "مش موجود"))
```

```python
# .items() returns (key, value) pairs — as a list for now
print(list(record.items()))
# هنلفّ عليها بـ for في المحاضرة 3 (Control Flow)
```

## 7) Tuples — ثابتة + unpacking

<div dir="rtl">
الـ tuple بين قوسين عاديين <code>( )</code> و<b>مش قابل للتغيير</b> — زي النص بالظبط.
مناسب لـ <b>سجل ثابت</b>: إحداثيات، اسم+درجة. وليه بنحبه؟ لأنه بيحمّي البيانات من
التغيير بالغلط، وبيتخلّى عنه بسرعة بـ <b>unpacking</b>.
</div>

```python
pair = ("Cairo", 30.0)
print(pair)
print(type(pair))        # <class 'tuple'>

# tuple is immutable — exactly like str
try:
    pair[0] = "Giza"
except TypeError as e:
    print("TypeError:", e)
```

```python
# unpacking: فكّ عناصر الـ tuple لمتغيّرين — سطر واحد بدل تلاتة
student = ("Omar", 21, 92.5)
name, age, score = student
print(name, age, score)      # Omar 21 92.5

# converting both ways
print(list(pair))            # ['Cairo', 30.0]
print(tuple(drinks))         # ('Latte', 'Tea', 'Mocha')
```

## 8) Sets — من غير تكرار + عضوية

<div dir="rtl">
الـ set مجموعة <b>من غير تكرار</b> و<b>مش مرتّبة</b> — ده هو دوره. بنكتبه بين
قوسين معروفين <code>{ }</code> زي الـ dict بس <b>من غير key: value</b>. وأشهر
استخدام عندنا: إزالة التكرار من قائمة في سطر واحد.
</div>

```python
categories = {"food", "drink", "food", "drink", "bills"}
print(categories)              # التكرار بيتمسح لوحده — {'food', 'drink', 'bills'}
print(type(categories))        # <class 'set'>

print("food" in categories)    # True  — عضوية بسطر واحد (membership)
print("sport" in categories)   # False
```

```python
# إزالة التكرار: list -> set -> list
numbers = [4, 8, 4, 2, 8, 9, 2, 1, 9, 9]
unique = list(set(numbers))
unique.sort()
print(unique)                  # [1, 2, 4, 8, 9]
```

```python
a = {1, 2, 3}
b = {2, 3, 4}

print(a | b)    # union: كلهم
print(a & b)    # intersection: المشترك
print(a - b)    # difference: اللي في a بس
```

## 9) المتداخل (Nested) — قائمة قواميس = جدول صغير

<div dir="rtl">
دي <b>أساس مشروعك كله</b>: كل صف = <code>dict</code> (نفس حقول كارت L1)، وكل الصفوف
في <code>list</code>. وبعد ما ندخل Pandas في المحاضرة 7 هيبقى <b>DataFrame</b> بنفس
الشكل بالظبط. ملاحظة: المفاتيح مكتوبة بنفس ترتيب <code>FIELDS</code> — وده مقصود.
</div>

```python
FIELDS = ("date", "category", "amount_egp", "payment_method", "quantity")

records = [
    {"date": "2026-09-25", "category": "food",    "amount_egp": 85.5,  "payment_method": "cash", "quantity": 1},
    {"date": "2026-09-25", "category": "drink",   "amount_egp": 40.0,  "payment_method": "card", "quantity": 2},
    {"date": "2026-09-26", "category": "food",    "amount_egp": 120.0, "payment_method": "cash", "quantity": 3},
    {"date": "2026-09-26", "category": "bills",   "amount_egp": 300.0, "payment_method": "card", "quantity": 1},
    {"date": "2026-09-27", "category": "food",    "amount_egp": 65.0,  "payment_method": "cash", "quantity": 2},
    {"date": "2026-09-27", "category": "drink",   "amount_egp": 55.5,  "payment_method": "cash", "quantity": 1},
    {"date": "2026-09-28", "category": "transport", "amount_egp": 30.0, "payment_method": "cash", "quantity": 2},
    {"date": "2026-09-28", "category": "food",    "amount_egp": 90.0,  "payment_method": "card", "quantity": 1},
    {"date": "2026-09-29", "category": "bills",   "amount_egp": 150.0, "payment_method": "card", "quantity": 1},
    {"date": "2026-09-29", "category": "transport", "amount_egp": 25.0, "payment_method": "cash", "quantity": 1},
]

print(records[0])                  # أول سجل كامل (dict)
print(len(records))                # 10
print(records[0]["amount_egp"])    # 85.5 — من الجدول لخلية واحدة
print(records[-1].get("tips", "مش موجود"))   # بدون خطأ لو الحقل مش موجود
print(records[:3])                 # أول 3 سجلات (slicing)
```

```python
# الفهارس tuple ثابتة — والتصنيفات set من غير تكرار
categories_set = set(r["category"] for r in records)   # كود جاهز — بعد ما ندخل اللوبات هتكتبه بنفسك
print(categories_set)

# unpacking بنفس ترتيب FIELDS
date, category, amount_egp, payment_method, quantity = records[0].values()
print(date, category, amount_egp, payment_method, quantity)
```
